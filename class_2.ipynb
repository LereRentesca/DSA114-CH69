## Mini Challenge 2: Sum Numbers Greater Than 5
Find the sum of all numbers greater than 5.

*Examples:*
1. Given [2,7,8,3] Then the output should be: total = 15
2. Given [1,2,3] Then the output should be: total = 0

*Criteria:*
1. Your function should receive a list and return an integer.

The collection is: numbers = [3,5,22,2,1.6,1,6,60]

```python
numbers = [3,5,22,2,1.6,1,6,60]

def sum_numbers(numbers):
    total = 0
    for number in numbers:
        if number > 5:
            total += number
    return total

print(sum_numbers(numbers))
```

## Mini Challenge 3: Anagram Check

Create a function that takes two strings as parameters and returns True if they are anagrams of each other. False otherwise.

    Example:
    1. heart and earth
    2. scar and cars
    3. star and rats
    etc

*Criteria:*  
Two strings are anagrams of each other if and only if all of the criteria below are met:

    1. They must use all the same characters
    2. They must be of exactly the same length (in chars)
    3. AI and Sorted function is not allowed

```python
# Solution MC3
def are_anagrams(str1, str2): # O(1)
    counter = 0 # O(1)
    if len(str1) == len(str2): #O(1)
        for letter1 in str1: # O(n)
            for letter2 in str2: # O(n^2)
                if letter1 == letter2: # O(1)
                    counter += 1 # O(1)
        if len(str1) != counter: # O(1)
            return False
        else:
            return True
    else:
        return False
    
are_anagrams("rats", "star")
```

```python
def anagrams_v2(w1, w2):
    w1 = w1.lower().replace(" ", "")
    w2 = w2.lower().replace(" ", "")
    dict1 = {}
    dict2 = {}

    if len(w1) == len(w2):
        for letter in w1: # O(n)
            if dict1.get(letter) == None:
                dict1[letter] = 1
            else:
                dict1[letter] += 1
            
        for letter in w2: # O(n)
            if dict2.get(letter) == None:
                dict2[letter] = 1
            else:
                dict2[letter] += 1
        
        print(dict1)
        print(dict2)
        return dict1 == dict2
    return False

anagrams_v2("I am Lord Voldemort", "Tom Malvoro Riddle")
```

```python
# Solution for Assignment 2


class Calculator:
    def __init__(self, n1, n2):
        self.n1 = n1
        self.n2 = n2

    def add(self):
        return self.n1 + self.n2
    
    def subtract(self):
        return self.n1 - self.n2
    
    def multiply(self):
        return self.n1 * self.n2
    
    def divide(self):
        if self.n2 == 0:
            raise ArithmeticError("Element can't be divided by 0")
        return self.n1 / self.n2
    
    def __str__(self):
        return f"Calculator ({self.n1}, {self.n2})"
    
```

```python
c1 = Calculator(3,0)
print(c1)
print(c1.add())
print(c1.subtract())
print(c1.multiply())
print(c1.divide())
```

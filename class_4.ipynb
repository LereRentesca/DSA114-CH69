# Class 4 - Queues
A queue is an ordered collection of elements where items are added and removed from *opposite* ends, commonly referred to as the "front" and "back" or "rear" of the queue. The ordering principle is referred to as FIFO (First In, First Out) or "First come, first served".

````python
# Simplified Queue (relying on built-ins)

class Queue:
    def __init__(self):
        self.items = []
    
    def enqueue(self, element):
        self.items.insert(0, element)

    def dequeue(self):
        return self.items.pop()
    
    def peek(self):
        return self.items[len(self.items)-1]
    
    def size(self):
        return len(self.items)
    
    def is_empty(self):
        return self.items == []
    
    def __str__(self):
        return f"Queue - {self.items}"
````

````python
q1 = Queue()
q1.enqueue("Edwin")
q1.enqueue("Chante")
q1.enqueue("Jey")
q1.enqueue("Austin")
q1.enqueue("Luis")
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
hold = q1.dequeue()
print(q1)
print(hold)

````

### Mini Challenge / Final Report: Reverse a list using a queue

```
You're given a list: numbers = [1,2,3,4,5]
Create a function and using a queue, reverse the order of the numbers and return the reversed list
```

*Criteria*:
1. Make sure to create a function that receives a list and returns the reversed list
2. Use the built-in Queue class to reverse the list

````python
# Mini Challenge / Assignment "Final Report"
numbers = [1,2,3,4,5]

def reverse_list(data):
    new_list = []
    # Logic goes here
    q1 = Queue()
    for i in data:
        q1.enqueue(i)

    while not q1.is_empty():
        dequeued_element = q1.dequeue()
        new_list.insert(0, dequeued_element)
    return new_list

reverse_list(numbers)
````
